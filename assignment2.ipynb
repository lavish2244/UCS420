3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate). 

i. Generate a list of 100 random numbers between 100 and 800 (inclusive). 

```python
import random
r_no = 1024170441
random.seed(r_no)

rand_list = [random.randint(100, 800) for _ in range(100)]
```

ii.  Count and print all odd numbers in the list. 

```python
odd = [n for n in rand_list if n % 2 != 0]

print("Odd numbers:", odd)
print("Number of odd numbers:", len(odd))

```

iii. Count and print all even numbers in the list. 

```python
even = [n for n in rand_list if n % 2 == 0]
print("Even numbers : ",even)
print("Number of Even numbers:", len(even))
```

iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension. 

```python
def prime(num):
    if num <= 1: return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0: return False
    return True

primes = [n for n in rand_list if prime(n)]
print("Number of prime numbers:", len(primes))
print("Prime Number :", primes)
```

v. Print the number that occurs most frequently in your list, and how many times it occurs. 

```python
freq = max(set(rand_list), key=rand_list.count)
count = rand_list.count(freq)
print("Most frequent number:", freq, "(appears", count, "times)")
```
