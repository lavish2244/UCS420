### Q.2 Questions on Basic NumPy Array:


```python
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print(f"Reversed Array: {reversed_arr}")

def find_most_frequent(array):
    vals, counts = np.unique(array, return_counts=True)
    max_freq = np.max(counts)
    most_freq_vals = vals[counts == max_freq]
    
    results = {}
    for val in most_freq_vals:
        indices = np.where(array == val)[0]
        results[val] = indices.tolist()
    return max_freq, results

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

freq_x, indices_x = find_most_frequent(x)
print(f"Array x most frequent value(s) and indices: {indices_x} with count {freq_x}")

freq_y, indices_y = find_most_frequent(y)
print(f"Array y most frequent value(s) and indices: {indices_y} with count {freq_y}")
```

### Q.4 

```python
arr_name = np.linspace(10, 100, 25)

print(f"Array: {arr_name}\n")
print(f"Dimensions (ndim): {arr_name.ndim}")
print(f"Shape: {arr_name.shape}")
print(f"Total elements (size): {arr_name.size}")
print(f"Data type (dtype): {arr_name.dtype}")
print(f"Total bytes consumed (nbytes): {arr_name.nbytes} bytes")

transposed_reshape = arr_name.reshape(25, 1)
print(f"\nTranspose via reshape (25, 1):\n{transposed_reshape}")
```

```python
ucs420_student = np.array([[10, 20, 30, 40], 
                           [50, 60, 70, 80], 
                           [90, 15, 20, 35]])

print(f"Original Array:\n{ucs420_student}\n")
print(f"Mean: {np.mean(ucs420_student)}")
print(f"Median: {np.median(ucs420_student)}")
print(f"Max: {np.max(ucs420_student)}")
print(f"Min: {np.min(ucs420_student)}")
print(f"Unique elements: {np.unique(ucs420_student)}")

reshaped_ucs420_student = ucs420_student.reshape(4, 3)
print(f"\nReshaped Array (4x3):\n{reshaped_ucs420_student}")

resized_ucs420_student = np.resize(ucs420_student, (2, 3))
print(f"\nResized Array (2x3):\n{resized_ucs420_student}")
```
