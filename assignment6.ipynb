## Q1. Sensor Readings

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("a.", corrected_temperature)

fahrenheit = (9/5) * temperature + 32
print("b.", fahrenheit)

readings_greater_32 = temperature[temperature > 32]
print("c.", readings_greater_32)

count_greater_32 = np.sum(temperature > 32)
print("d.", count_greater_32)

print("e. Vectorization performs operations on the whole NumPy array without an explicit loop.")
print("   Boolean indexing selects values satisfying a condition. Both are concise and efficient.")
```

## Q2. Daily Steps

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("a. Total steps:", np.sum(steps))

print("b. Mean steps:", np.mean(steps))

print("c. Maximum:", np.max(steps))
print("   Minimum:", np.min(steps))

print("d. Total for each day:", np.sum(steps, axis=0))

print("e. Total for each user:", np.sum(steps, axis=1))

print("f. Position of maximum (user, day):",
      np.unravel_index(np.argmax(steps), steps.shape))
```

## Q3. NumPy Arrays, Slicing, Copying and Reshaping

```python
original = np.array([1, 2, 3, 4, 5, 6])
print("a. Original:", original)

subset = original[1:4]
print("b. Subset:", subset)

subset[0] = 999
print("c. Original:", original)
print("   Subset:", subset)

copied = original[1:4].copy()
copied[0] = 500
print("d. Original:", original)
print("   Copied:", copied)

matrix = np.arange(1, 13).reshape(3, 4)
print("e. Matrix:")
print(matrix)

print("f. First row:", matrix[0])
print("   Last row:", matrix[-1])
print("   Second column:", matrix[:, 1])
print("   Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

flattened = matrix.flatten()
raveled = matrix.ravel()
print("g. flatten():", flattened)
print("   ravel():", raveled)

raveled[0] = 1000
print("h. Matrix after modifying ravel():")
print(matrix)

flattened[1] = 2000
print("i. Matrix after modifying flatten():")
print(matrix)

print("j. Shape:", matrix.shape)
print("   ndim:", matrix.ndim)
print("   size:", matrix.size)
print("   dtype:", matrix.dtype)
```

## Q4. Cognitive-Assistive System and OLS

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("a. Shape:", X.shape)
print("   Dimensions:", X.ndim)

X_T = X.T
print("\nb. X.T:")
print(X_T)
print("   Transpose changes rows into columns and columns into rows.")

XTX = X.T @ X
print("\nc. X.T @ X:")
print(XTX)

XTX_inv = np.linalg.inv(XTX)
print("\nd. Inverse of X.T @ X:")
print(XTX_inv)

beta = XTX_inv @ X.T @ y
print("\ne. Coefficients beta:")
print(beta)

print("\nf. beta[0] = Sleep Hours coefficient")
print("   beta[1] = Activity Level coefficient")
print("   beta[2] = Stress Level coefficient")
print("   Each coefficient estimates the change in assistance score for")
print("   a one-unit increase in that feature, keeping the others fixed.")

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("\ng. New user:", new_user)
print("   Predicted assistance score:", predicted_score)
```
